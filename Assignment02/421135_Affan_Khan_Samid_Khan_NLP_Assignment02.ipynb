This notebook processes eight authored sample sentences each in English, Hindi and Marathi. It performs Unicode normalization, sentence segmentation, regex tokenization, curated stop-word removal, English stemming with NLTK Snowball, and word-frequency analysis. Hindi and Marathi stemming is not applied because a validated language-specific stemmer is not configured.

Run from the Assignment02 folder, keeping the accompanying `.py` source beside this notebook.

```python
from pathlib import Path
import sys, importlib.util
PROJECT_DIR = Path.cwd()
SOURCE = PROJECT_DIR / '421135_Affan_Khan_Samid_Khan_NLP_Assignment02.py'
if not SOURCE.exists():
    raise FileNotFoundError('Run this notebook from the Assignment02 directory with the .py file beside it.')
spec = importlib.util.spec_from_file_location('assignment02', SOURCE)
assignment02 = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = assignment02
spec.loader.exec_module(assignment02)
OUTPUT_DIR = PROJECT_DIR / 'outputs'
comparison, frequencies = assignment02.run(OUTPUT_DIR)
for row in comparison:
    print(f"{row['language']}: sentences={row['sentences']}, tokens={row['tokens']}, unique={row['unique_tokens']}, stop words removed={row['stopwords_removed']}, remaining={row['tokens_after_stopword_removal']}")
```

```python
for language in assignment02.CORPUS:
    top = [r for r in frequencies if r['language'] == language][:5]
    print(language + ' top frequency items: ' + ', '.join(f\"{r['token']} ({r['frequency']})\" for r in top))
```

## Interpretation and limitations

- The corpus is small and authored for this assignment; statistics describe these sentences only, not each language in general.
- Stop-word lists are compact teaching lists and may remove useful words depending on the task.
- English uses NLTK Snowball stemming. Hindi and Marathi stemming is not applied rather than claiming an unvalidated morphological method.
- The regex tokenizer is transparent but is not a full linguistic tokenizer, POS tagger, or named-entity recognition model.
