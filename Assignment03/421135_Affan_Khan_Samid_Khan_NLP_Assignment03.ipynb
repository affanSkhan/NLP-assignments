## Technique
Supervised first-order Hidden Markov Model (HMM), add-alpha smoothing (alpha=0.1), suffix/shape fallback for out-of-vocabulary tokens, and Viterbi decoding in log space.

The included `tagged_corpus.csv` contains 48 manually tagged training sentences and 17 gold-labeled held-out sentences (100 tokens). This is a small instructional corpus, not a benchmark.

```python
from pathlib import Path
import importlib.util, sys
source = Path('421135_Affan_Khan_Samid_Khan_NLP_Assignment03.py')
spec = importlib.util.spec_from_file_location('pos_assignment03', source)
pos = importlib.util.module_from_spec(spec)
sys.modules[spec.name] = pos
spec.loader.exec_module(pos)
model, metrics, per_tag, confusion, predictions = pos.run('outputs')
for key in ['test_sentences','test_tokens','correct_tokens','accuracy','macro_precision','macro_recall','macro_f1','majority_baseline_accuracy','word_majority_baseline_accuracy']:
    print(f'{key}: {metrics[key]}')
```

```python
for sentence in ['The new model predicts useful results.', 'They watch the movie.']:
    words = pos.TOKEN_RE.findall(sentence)
    print(sentence)
    print(' '.join(f'{word}/{tag}' for word, tag in model.tag(words)))
```

## Interpretation

The HMM achieves 89.00% token accuracy (macro-F1 0.8744) on this held-out sample. The global majority-tag baseline is 26.00%; the word-majority baseline is 92.00%, which is slightly higher because the sample is lexically predictable. The results are a demonstration of the algorithm, not a benchmark claim.

Common errors involve category ambiguity (e.g., *present* or *light*) and rare suffix forms (e.g., *-ing*). See `outputs/sentence_predictions.csv`, `outputs/per_tag_metrics.csv`, `outputs/confusion_matrix.csv`, `outputs/metrics.json`, and `outputs/results.txt` after running the script.
