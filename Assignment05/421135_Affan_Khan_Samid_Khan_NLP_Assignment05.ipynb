## Research paper

Hutto & Gilbert (2014), *VADER: A Parsimonious Rule-Based Model for Sentiment Analysis of Social Media Text*, ICWSM 8(1), 216-225. DOI: https://doi.org/10.1609/icwsm.v8i1.14550.

VADER uses a human-validated lexicon and five heuristics: punctuation, capitalization, degree modifiers, contrastive `but`, and negation in the preceding three-token window. The compound score is thresholded at +/-0.05. Neutral predictions count as misses against the binary positive/negative gold labels and are reported separately.

Important difference: the paper's published F1 is from its own evaluation data and protocol. It is not the result of this assignment experiment.
